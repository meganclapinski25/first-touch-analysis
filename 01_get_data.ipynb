```python
#Imports
from statsbombpy import sb 
import warnings
import os
import glob
import pandas as pd
warnings.filterwarnings("ignore", message="credentials were not supplied")
```

```python
matches = sb.matches(competition_id=49, season_id=107)
print(len(matches))

os.makedirs("data/events", exist_ok = True)
matches.to_pickle("data/matches.pl1")
```

```python
events = sb.events(match_id=matches.match_id.iloc[0])
print(events.shape)
events["type"].value_counts().head(10)
```

```python
#download every match in the dataframe
for i, match_id in enumerate(matches.match_id):
    path = f"data/events/{match_id}.pkl"
    if os.path.exists(path):
        continue
    try:
        ev = sb.events(match_id=match_id)
        ev.to_pickle(path)
        print(f"{i+1}/{len(matches)} saved")
    except Exception as e:
        print(f"failed {match_id}: {e}")
```

```python
files = glob.glob("data/events/*.pkl")
all_events = pd.concat(
    [pd.read_pickle(f) for f in files], ignore_index=True
)
all_events.to_pickle("data/nwsl_2023_events.pkl")
print(len(files), all_events.shape)
```
