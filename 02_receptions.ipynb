```python
import pandas as pd

ev = pd.read_pickle("data/nwsl_2023_events.pkl")
ev = ev.sort_values(["match_id", "index"]).reset_index(drop=True)
ev["t"] = pd.to_timedelta(ev["timestamp"]).dt.total_seconds()
```

```python
cols = ["id", "match_id", "index", "period", "t", "minute",
        "team", "player", "position", "location",
        "under_pressure", "possession", "ball_receipt_outcome"]

reps = ev[ev["type"] =="Ball Receipt*"][cols].copy()
reps["under_pressure"] = reps["under_pressure"].eq(True)
reps.to_pickle("data/receptions.pkl")
reps.shape
```

```python
## Data Dictionary for positions 

role_map = {
     # Goalkeeper
    "Goalkeeper": "GK",

    # Center backs
    "Right Center Back": "CB",
    "Center Back": "CB",
    "Left Center Back": "CB",

    # Fullbacks / wing backs
    "Right Back": "FB",
    "Left Back": "FB",
    "Right Wing Back": "FB",
    "Left Wing Back": "FB",

    # Central midfielders
    "Center Defensive Midfield": "CM",
    "Right Defensive Midfield": "CM",
    "Left Defensive Midfield": "CM",
    "Right Center Midfield": "CM",
    "Left Center Midfield": "CM",

    # Wingers / attacking midfielders
    "Right Wing": "W",
    "Left Wing": "W",
    "Right Midfield": "W",
    "Left Midfield": "W",
    "Center Attacking Midfield": "W",
    "Right Attacking Midfield": "W",
    "Left Attacking Midfield": "W",

    # Strikers
    "Center Forward": "ST",
    "Right Center Forward": "ST",
    "Left Center Forward": "ST",
}

reps["role"] = reps["position"].map(role_map)
```
